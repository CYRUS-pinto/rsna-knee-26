# RSNA Knee Engine 10 Fold 1: Swin-Base-384 2.5D MIL Master (A100 GPU)

**Key Architectural Innovations:**
- **Hierarchical Vision Transformer:** `swin_base_patch4_window12_384.ms_in22k_ft_in1k` providing pure non-convolutional global attention.
- **Zero Inductive Bias Overlap:** Completely orthogonal feature representations from ConvNeXt.
- **Fold 1 Training:** Disjoint validation split (`val_ids = shuffled_train[n_val : 2*n_val]`), completing 2-fold Swin representation.
- **Asymmetric Focal Loss (gamma=1.5):** Suppresses easy negative slices and targets rare pathologies.
- **Target Gold AUC:** >= 0.7800

```python
# 3. Write Engine 10 Swin-Base-384 Training Script
with open('train_engine10_swin.py', 'w', encoding='utf-8') as f:
    f.write('''"""
RSNA Knee Abnormality Detection — Engine 10: Swin-Base-384 2.5D MIL Master
Autonomous GPU Training Pipeline with Asymmetric Focal Loss & 384px Vision Transformer
"""

import os, sys, time, math, random
import numpy as np, pandas as pd
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import timm

TARGETS = [
    'ACL', 'MCL', 'Medial Meniscus', 'Lateral Meniscus',
    'Medial OA', 'Lateral OA', 'PF OA', 'Effusion',
    'Synovitis', "Baker's", 'Contusion', 'Fracture'
]

def fast_roc_auc(y_true, y_score):
    y_true = np.asarray(y_true).astype(bool)
    y_score = np.asarray(y_score)
    n_pos = np.sum(y_true)
    n_neg = len(y_true) - n_pos
    if n_pos == 0 or n_neg == 0:
        return np.nan
    ranks = np.empty(len(y_score), dtype=float)
    order = np.argsort(y_score)
    ranks[order] = np.arange(len(y_score))
    _, inv, counts = np.unique(y_score, return_inverse=True, return_counts=True)
    if np.any(counts > 1):
        tie_ranks = np.zeros(len(counts))
        np.add.at(tie_ranks, inv, ranks)
        tie_ranks = tie_ranks / counts
        ranks = tie_ranks[inv]
    rank_sum = np.sum(ranks[y_true])
    u = rank_sum - (n_pos * (n_pos - 1)) / 2.0
    return float(u / (n_pos * n_neg))

class TwoPartVols:
    def __init__(self, a, b):
        self.a, self.b = a, b
        self.n_a = a.shape[0]
        self.shape = (a.shape[0] + (b.shape[0] if b is not None else 0),) + tuple(a.shape[1:])
    def __len__(self): return self.shape[0]
    def __getitem__(self, row):
        return self.a[row] if row < self.n_a else self.b[row - self.n_a]

class SwinMILDataset(Dataset):
    def __init__(self, ids, id2row, labels, vols, res=384, train=True, n_slices=6):
        self.ids = ids
        self.id2row = id2row
        self.labels = labels
        self.vols = vols
        self.res = res
        self.train = train
        self.n_slices = n_slices

    def __len__(self):
        return len(self.ids)

    def __getitem__(self, i):
        uid = self.ids[i]
        row = self.id2row[uid]
        vol = self.vols[row]  # shape: (D, H, W)
        D = vol.shape[0]

        # Sample n_slices centers spaced across depth, each taking [c-1, c, c+1] 3-channel triplet
        if D <= 4:
            centers = [min(D - 1, max(1, k)) for k in range(self.n_slices)]
        else:
            step = (D - 2) / float(self.n_slices + 1)
            if self.train:
                # Add slight random jitter during training
                jitter = (random.random() - 0.5) * step * 0.5
                centers = [int(round(1 + (k + 1) * step + jitter)) for k in range(self.n_slices)]
            else:
                centers = [int(round(1 + (k + 1) * step)) for k in range(self.n_slices)]
            centers = [max(1, min(D - 2, c)) for c in centers]

        triplets = []
        for c in centers:
            trip = vol[c-1:c+2].astype(np.float32) / 255.0  # (3, H, W)
            if trip.shape[0] < 3:
                trip = np.repeat(vol[c:c+1].astype(np.float32) / 255.0, 3, axis=0)
            triplets.append(trip)

        stack = np.stack(triplets, axis=0)  # (n_slices, 3, H, W)
        t = torch.from_numpy(np.ascontiguousarray(stack))

        if t.shape[-1] != self.res or t.shape[-2] != self.res:
            # Bilinear resize to 384x384
            t = F.interpolate(t, size=(self.res, self.res), mode='bilinear', align_corners=False)

        if self.train:
            if random.random() < 0.5:
                t = torch.flip(t, dims=[-1])  # Horizontal flip
            gain = 1.0 + (random.random() - 0.5) * 0.12
            t = (t * gain).clamp(0, 1)

        # Normalize with ImageNet standard mean and std per slice
        mean = torch.tensor([0.485, 0.456, 0.406]).view(1, 3, 1, 1)
        std = torch.tensor([0.229, 0.224, 0.225]).view(1, 3, 1, 1)
        t = (t - mean) / std

        y = torch.tensor(self.labels[uid], dtype=torch.float32)
        return t, y

class SwinMILModel(nn.Module):
    def __init__(self, model_name='swin_base_patch4_window12_384.ms_in22k_ft_in1k', num_classes=12):
        super().__init__()
        self.backbone = timm.create_model(model_name, pretrained=True, num_classes=0)
        feat_dim = self.backbone.num_features
        # Attention MIL Pooling
        self.attention = nn.Sequential(
            nn.Linear(feat_dim, 128),
            nn.Tanh(),
            nn.Linear(128, 1)
        )
        self.head = nn.Linear(feat_dim, num_classes)

    def forward(self, x):
        # x shape: (B, S, 3, 384, 384)
        B, S, C, H, W = x.shape
        x = x.view(B * S, C, H, W)
        feats = self.backbone(x)  # (B*S, feat_dim)
        feats = feats.view(B, S, -1)

        # Compute attention weights across S slices
        attn_scores = self.attention(feats)  # (B, S, 1)
        attn_weights = F.softmax(attn_scores, dim=1)  # (B, S, 1)

        # Weighted MIL aggregation + max-pool skip
        pooled_attn = torch.sum(feats * attn_weights, dim=1)  # (B, feat_dim)
        pooled_max, _ = torch.max(feats, dim=1)  # (B, feat_dim)
        pooled = 0.7 * pooled_attn + 0.3 * pooled_max

        logits = self.head(pooled)
        return logits

class AsymmetricFocalLoss(nn.Module):
    def __init__(self, gamma=1.5, eps=1e-7):
        super().__init__()
        self.gamma = gamma
        self.eps = eps

    def forward(self, logits, targets):
        p = torch.sigmoid(logits)
        p = torch.clamp(p, self.eps, 1.0 - self.eps)
        pos_loss = - targets * torch.pow(1.0 - p, self.gamma) * torch.log(p)
        neg_loss = - (1.0 - targets) * torch.pow(p, self.gamma) * torch.log(1.0 - p)
        loss = pos_loss + neg_loss
        return loss.mean()

def main():
    print('=== Starting RSNA Knee Engine 10: Swin-Base-384 2.5D MIL Master ===', flush=True)
    device = torch.device('cuda:0' if torch.cuda.is_available() else 'cpu')
    print(f'Device: {device}', flush=True)

    # Load Corpus
    va = np.load('/content/corpus/all_vols.npy', mmap_mode='r')
    vb = np.load('/content/corpus/extra_vols.npy', mmap_mode='r') if os.path.exists('/content/corpus/extra_vols.npy') else None
    vols = TwoPartVols(va, vb)

    a_ids = np.load('/content/corpus/all_ids.npy', allow_pickle=True).astype(str)
    b_ids = np.load('/content/corpus/extra_ids.npy', allow_pickle=True).astype(str) if os.path.exists('/content/corpus/extra_ids.npy') else np.array([])
    all_ids = np.concatenate([a_ids, b_ids]) if len(b_ids) > 0 else a_ids

    id2row = {u: i for i, u in enumerate(all_ids)}
    idset = set(all_ids)
    print(f'Loaded corpus: {len(all_ids)} examinations.', flush=True)

    # Load Gold Anchor & Continuous Soft Labels
    tr = pd.read_csv('/content/rsna_comp/train.csv')
    tr['StudyInstanceUID'] = tr['StudyInstanceUID'].astype(str)
    gold_df = tr[tr[TARGETS].notna().all(axis=1)].copy().set_index('StudyInstanceUID')
    gold_ids = [u for u in gold_df.index if u in idset]
    print(f'Anchored on {len(gold_ids)} official radiologist gold studies.', flush=True)

    soft_path = '/content/labels/SUPER_CONSENSUS_labels.parquet'
    soft = pd.read_parquet(soft_path).set_index('StudyInstanceUID')

    train_ids = [u for u in all_ids if u in soft.index and u not in set(gold_ids)]
    labels = {u: soft.loc[u, TARGETS].values.astype(np.float32) for u in train_ids}
    for u in gold_ids:
        labels[u] = gold_df.loc[u, TARGETS].values.astype(np.float32)

    # 5-fold split: FOLD 1
    np.random.seed(42)
    shuffled_train = list(train_ids)
    np.random.shuffle(shuffled_train)
    n_val = len(shuffled_train) // 5
    val_ids = shuffled_train[n_val:2*n_val]
    core_train_ids = shuffled_train[:n_val] + shuffled_train[2*n_val:]

    tds = SwinMILDataset(core_train_ids, id2row, labels, vols, res=384, train=True, n_slices=6)
    vds = SwinMILDataset(val_ids, id2row, labels, vols, res=384, train=False, n_slices=6)
    gds = SwinMILDataset(gold_ids, id2row, labels, vols, res=384, train=False, n_slices=6)

    # Batch size 8 on A100 SXM4 (8 * 6 slices = 48 forward passes per batch)
    tl = DataLoader(tds, batch_size=8, shuffle=True, num_workers=4, pin_memory=True, drop_last=True)
    vl = DataLoader(vds, batch_size=16, shuffle=False, num_workers=4, pin_memory=True)
    gl = DataLoader(gds, batch_size=16, shuffle=False, num_workers=4, pin_memory=True)

    print('Initializing Swin-Base-384 Vision Transformer for 384px MIL...', flush=True)
    model = SwinMILModel('swin_base_patch4_window12_384.ms_in22k_ft_in1k', num_classes=len(TARGETS)).to(device)

    epochs = 8
    opt = torch.optim.AdamW(model.parameters(), lr=1.5e-4, weight_decay=1e-3)
    sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=epochs * len(tl), eta_min=1e-6)
    lossf = AsymmetricFocalLoss(gamma=1.5)
    scaler = torch.amp.GradScaler('cuda')

    best_gold_auc = 0.0
    t0 = time.time()

    for ep in range(epochs):
        model.train()
        tot = 0.0
        for x, y in tl:
            x, y = x.to(device, non_blocking=True), y.to(device, non_blocking=True)
            opt.zero_grad(set_to_none=True)
            with torch.amp.autocast('cuda'):
                logits = model(x)
                loss = lossf(logits, y)
            scaler.scale(loss).backward()
            scaler.unscale_(opt)
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            scaler.step(opt)
            scaler.update()
            sched.step()
            tot += loss.item()

        # Gold Evaluation
        model.eval()
        P, Y = [], []
        with torch.no_grad():
            for gx, gy in gl:
                with torch.amp.autocast('cuda'):
                    gl_out = model(gx.to(device))
                P.append(torch.sigmoid(gl_out).float().cpu().numpy())
                Y.append(gy.numpy())
        P = np.concatenate(P, axis=0)
        Y = np.concatenate(Y, axis=0)
        gold_aucs = [fast_roc_auc(Y[:, j], P[:, j]) for j in range(len(TARGETS))]
        macro_gold = float(np.nanmean(gold_aucs))

        print(f'Epoch {ep+1:2d}/{epochs} | Loss: {tot/len(tl):.4f} | Gold-58 Macro AUC: {macro_gold:.4f} (Best: {max(best_gold_auc, macro_gold):.4f}) | {time.time()-t0:.0f}s', flush=True)
        for target_name, target_auc in zip(TARGETS, gold_aucs):
            print(f'   {target_name:18s}: {target_auc:.4f}', flush=True)

        if macro_gold > best_gold_auc:
            best_gold_auc = macro_gold
            torch.save({
                'epoch': ep + 1,
                'model': model.state_dict(),
                'gold_auc': macro_gold,
                'aucs': dict(zip(TARGETS, gold_aucs)),
                'arch': 'swin_base_patch4_window12_384',
                'targets': TARGETS
            }, '/content/engine10_swin_384_best.pt')
            print(f'  -> New Best Saved with Gold AUC: {macro_gold:.4f}', flush=True)

    print(f'\\nTraining Finished! Peak Gold-58 Macro-AUC: {best_gold_auc:.4f}', flush=True)

if __name__ == '__main__':
    main()
''')
print('Wrote train_engine10_swin.py successfully!')
```
